```python
# Importando os dados para o teste de  hipóteses
df_loop_ohare = pd.read_csv('/datasets/project_sql_result_07.csv')

# Imprimindo resultados
print(df_loop_ohare.head())
print(df_loop_ohare.info())
```

```python
# Adicionando uma coluna indicando se é sábado chuvoso
df_loop_ohare['is_saturday_rain'] = (
    (df_loop_ohare['weekday'] == 'Saturday') &
    (df_loop_ohare['weather_conditions'].str.contains('Bad', case=False, na=False))
)

print(df_loop_ohare)
```

```python
from scipy.stats import mannwhitneyu

# duas amostras
grupo1 = df_loop_ohare[df_loop_ohare['is_saturday_rain']]['duration_seconds'].dropna()
grupo2 = df_loop_ohare[~df_loop_ohare['is_saturday_rain']]['duration_seconds'].dropna()

# nível de significância
alpha = 0.05

print(f"Tamanho do grupo com chuva: {len(grupo1)}")
print(f"Tamanho do grupo sem chuva: {len(grupo2)}")

if len(grupo1) == 0 or len(grupo2) == 0:
    print("Não há dados suficientes em um dos grupos para realizar o teste.")
else:
    # teste
    stat, p = mannwhitneyu(grupo1, grupo2, alternative='two-sided')

    print(f"Estatística U={stat:.2f}, p-valor={p:.4f}")

    if p < alpha:
        print("Rejeitamos H₀: a duração média é diferente aos sábados chuvosos.")
    else:
        print("Não rejeitamos H₀: não há evidências de que a duração média mude aos sábados chuvosos.")
     
```

# Teste de Hipóteses: Duração das corridas aos sábados com e sem chuva

## Formulação das Hipóteses

O objetivo deste teste era verificar se existe diferença significativa na **duração média das corridas** aos sábados, comparando dias com chuva e dias sem chuva.

- **Hipótese nula (H₀):**
  A duração média das corridas aos sábados com chuva é **igual** à duração média aos sábados sem chuva.
  
  Em termos matemáticos:  
  μ_chuva = μ_sem_chuva

- **Hipótese alternativa (H₁):**
  A duração média das corridas aos sábados com chuva é **diferente** da duração média aos sábados sem chuva.
  
  Em termos matemáticos:  
  μ_chuva ≠ μ_sem_chuva

Como queremos verificar se há qualquer diferença (para mais ou para menos), trata-se de um **teste bicaudal**.

---

## Nível de significância

Definimos o nível de significância como:  
α = 0.05

Ou seja, aceitaremos um risco de 5% de rejeitar a hipótese nula quando ela é verdadeira.

---

## Critério do teste

Para comparar as médias de duas amostras independentes (corridas aos sábados com chuva vs. sem chuva), utilizamos o seguinte critério:

- Como as amostras são suficientemente grandes e não assumimos normalidade dos dados, aplicamos o teste **Mann-Whitney U**, que é um teste não-paramétrico.
- O teste Mann-Whitney é adequado para comparar duas amostras independentes quando não podemos assumir que os dados seguem distribuição normal ou quando os grupos têm tamanhos diferentes.

O critério para decisão é:

- Se o valor-p < 0.05 → Rejeitamos H₀ → Existe diferença significativa na duração média.
- Caso contrário → Não rejeitamos H₀ → Não há evidências de diferença significativa.

---

## Resultados

- Número de corridas aos sábados com chuva: **180**
- Número de corridas aos sábados sem chuva: **888**
- Estatística U do teste: **108193.50**
- Valor-p do teste: **< 0.0001**

---

## Conclusão

Como o valor-p encontrado é menor que o nível de significância adotado (0.05), **rejeitamos a hipótese nula**.

Há evidências estatísticas suficientes para afirmar que a duração média das corridas aos sábados com chuva é significativamente diferente da duração média aos sábados sem chuva.
